```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv("preds/metrics_summary.csv", index_col=0)
```

```python
df
```

```python
df_stab = df[["Stability_baseline", "Stability_v2", "Stability_v3"]]
df_stab.columns = ["baseline", "v2", "v3"]
df_stab = df_stab.melt(value_name="Stability", var_name="Method")

df_nme = df[["NME_baseline", "NME_v2", "NME_v3"]]
df_nme.columns = ["baseline", "v2", "v3"]
df_nme = df_nme.melt(value_name="NME", var_name="Method")
```

```python
sns.boxplot(df_stab, x="Stability", y="Method")
```

```python
sns.boxplot(df_nme, x="NME", y="Method")
```
